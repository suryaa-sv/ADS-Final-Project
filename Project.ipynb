```python
import pandas as pd

# 1. Load the data
players = pd.read_csv(r'E:\ADS project\players.csv\players.csv')
appearances = pd.read_csv(r'E:\ADS project\appearances.csv\appearances.csv')
valuations = pd.read_csv(r'E:\ADS project\player_valuations.csv\player_valuations.csv')

# 2. Select only the columns we actually need (doing this step-by-step)
players_basic = players[['player_id', 'name', 'sub_position', 'date_of_birth']]
valuations_basic = valuations[['player_id', 'date', 'market_value_in_eur']]

# 3. Group the appearances data to get total stats per player
stats_grouped = appearances.groupby('player_id').sum()
stats_grouped = stats_grouped.reset_index()

# Keep only the important stat columns
stats_basic = stats_grouped[['player_id', 'goals', 'assists', 'minutes_played', 'yellow_cards', 'red_cards']]

# 4. Get the latest valuation for each player
# Sort by date so the newest date is at the top, then drop the duplicates
val_sorted = valuations_basic.sort_values(by='date', ascending=False)
latest_val = val_sorted.drop_duplicates(subset='player_id', keep='first')

# 5. Merge the dataframes one at a time
first_merge = pd.merge(players_basic, stats_basic, on='player_id')
final_data = pd.merge(first_merge, latest_val, on='player_id')

# 6. Drop any blank values in the market value column
final_data = final_data.dropna(subset=['market_value_in_eur'])

# 7. REDUCE THE DATASET TO 750 PLAYERS
# Sort by market value (highest to lowest) and just take the top 750 rows
final_data = final_data.sort_values(by='market_value_in_eur', ascending=False)
final_data = final_data.head(750)

# 8. Save the file
final_data.to_csv(r'E:\ADS project\powerbi_player_valuations.csv', index=False)
print("Data cleaned, reduced to 750 players, and saved!")
```
